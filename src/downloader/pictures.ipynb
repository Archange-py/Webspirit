```python
from downloader.tools.csvmanager import HyperLink
```

```python
import json

ntb_filename: str = 'pictures.ipynb'
i: int = 0

with open(ntb_filename, 'r', encoding='utf-8') as f:
    notebook = json.load(f)

cell: dict = notebook['cells'][i]
links: list[HyperLink] = [
        link.removesuffix('  \n') for link in cell['source']
]   if cell['cell_type'] in ('markdown', 'raw') else ['']

print(links)
```

```python
img_links: list[dict[str, str]] = [
    {
        'Default' : f"https://img.youtube.com/vi/{link.id}/0.jpg",
        'Medium' : f"https://img.youtube.com/vi/{link.id}/0.jpg",
        'High' : f"https://img.youtube.com/vi/{link.id}/0.jpg",
        'Standard' : f"https://img.youtube.com/vi/{link.id}/0.jpg",
        'Maximum' : f"https://img.youtube.com/vi/{link.id}/0.jpg"
    }
    for link in links
]

print(img_links)
```

```python
from IPython.display import display, HTML

html = "<div style='display: flex; flex-wrap: wrap; gap: 10px;'>"
for url in img_links:
    html += f"<img src='{url}' style='width: 150px; height: auto;'>"
html += "</div>"

display(HTML(html))
```
