## ErrorContextManager or *ecm*

```python
@ECM("Error in the divide function 1") # or ecm
def divide(a: int, b: int) -> float:
    return a / b

divide(10, 0)
```

## CheckType & ValidatePathOrUrl

```python
class BookOfLink:
    @CheckType
    def append_url1(self, url: HyperLink, nbr: int) -> tuple[HyperLink, int]:
        return type(url), type(nbr)

    @CheckType
    def append_url2(self, url: str, nbr: int | None = None) -> tuple[str, int | None]:
        return type(url), type(nbr)

    @CheckType
    def append_url3(self, url: HyperLink, nbr: int, string: str) -> tuple[HyperLink, int, str]:
        return type(url), type(nbr), type(string)

    @CheckType('path')
    def append_path(self, path: StrPath) -> StrPath:
        return type(path)

    @ValidatePathOrUrl('link')
    def append1(self, link: PathOrURL = "https://youtu.be/_0Pf48RqSsg") -> PathOrURL:
        return type(link)

    @ValidatePathOrUrl('url')
    def append2(self, url: HyperLink) -> HyperLink:
        return type(url)
```

```python
BookOfLink.append1.__qualname__
BookOfLink.append1.__code__
```

```python
@CheckType
def append_url1(url: HyperLink, nbr: int) -> tuple[HyperLink, int]:
    return type(url), type(nbr)

@CheckType
def append_url2(url: str, nbr: int | None = None) -> tuple[str, int | None]:
    return type(url), type(nbr)

@CheckType('url', 'nbr', 'string')
def append_url3(url: HyperLink, nbr: int, string: str) -> tuple[HyperLink, int, str]:
    return type(url), type(nbr), type(string)

@CheckType('path')
def append_path(path: StrPath) -> StrPath:
    return type(path)

@ValidatePathOrUrl('link')
def append1(link: PathOrURL = "https://youtu.be/_0Pf48RqSsg") -> PathOrURL:
    return type(link)

@ValidatePathOrUrl('url')
def append2(url: HyperLink) -> HyperLink:
    return type(url)
```

### CheckType

#### Simple Example

```python
@CheckType(return_=True)
def function_1(arg1: str, arg2: str) -> int:
    return arg1 + arg2

function_1('1', 0)
```

```python
@CheckType('arg2', return_=True)
def function_2(arg1, arg2: str) -> int:
    return arg1 + arg2

function_2('1', 0)
```

```python
@CheckType(convert=False)
def function_3(arg1: str, arg2: str) -> int:
    return arg1 + arg2

with ecm('An error was occurred in the function_3'):
    function_3('1', 0)
```

```python
@CheckType('arg1', 'a')
def function_4(arg1: str, arg2: str) -> int:
    return arg1 + arg2

with ecm('An error was occurred in the function_4'):
    function_4('1', '0')
```

```python
@CheckType
def function_5(arg1: str, arg2: str) -> str:
    return arg1 + arg2

function_5('1', 0)
```
